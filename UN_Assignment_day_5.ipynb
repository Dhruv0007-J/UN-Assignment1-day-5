# 3. STRING TO LIST
Create sent

```python
sent = "Coronavirus Caused Lockdowns Around The World"
```

Split into words

```python
words = sent.split()
print(words)
```

Convert words to lowercase

```python
words_lower = []

for word in words:
    words_lower.append(word.lower())

print(words_lower)
```

Check whether country exists

```python
print('country' in words_lower)
```

Remove the

```python
words_lower.remove('the')
print(words_lower)
```

Select first four words

```python
x4 = words_lower[:4]
print(x4)
```

Convert list into single string

```python
result = " ".join(words_lower)
print(result)
```

# 5. SET OPERATIONS
Create colors and fruits

```python
colors = ['red', 'blue', 'orange']
fruits = ['orange', 'grapes', 'apples']
```

Print both

```python
print(colors)
print(fruits)
```

Convert into sets

```python
colors_set = set(colors)
fruits_set = set(fruits)

print(colors_set)
print(fruits_set)
```

Union

```python
print(colors_set.union(fruits_set))
```

Intersection

```python
print(colors_set.intersection(fruits_set))
```

Fruits but not colors

```python
print(fruits_set.difference(colors_set))
```
